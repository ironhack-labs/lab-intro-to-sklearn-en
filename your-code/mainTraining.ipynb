# Challenge 3 - Peform Supervised Learning on a Pandas Dataframe

Now that we have looked at data that has been formatted for scikit-learn, let's look at data that we will need to format ourselves.

In the next cell, load the `auto-mpg.csv` file included in this folder and assign it to a variable called `auto`.

```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

csv_url = "https://raw.githubusercontent.com/ironhack-labs/lab-intro-to-sklearn-en/main/auto-mpg.csv"

auto = pd.read_csv(csv_url)

print("Datos cargados:", auto.shape)
```

Look at the first 5 rows using the `head()` function:

```python

auto.head()

```

Evaluate the data to ensure that all numeric columns are correctly detected as such by pandas. If a column is misclassified as object, coerce it to numeric.

```python

print("Tipos de datos antes de la conversión:")
print(auto.dtypes)

# Convertir las columnas numéricas
num_cols = [col for col in auto.columns if col != "car_name"]

for col in num_cols:
    auto[col] = pd.to_numeric(auto[col], errors="coerce")

print("\nTipos de datos después de la conversión:")
print(auto.dtypes)
```

What is the newest model year and the oldest model year?

```python

newest_year = auto["model_year"].max()
oldest_year = auto["model_year"].min()

print("Año más reciente:", newest_year)
print("Año más antiguo:", oldest_year)
```

Check the dataset for missing values and remove all rows containing at least one missing value.

```python

print("Valores ausentes antes de limpiar:")
print(auto.isna().sum())

# Guardar el número de filas originales
filas_antes = len(auto)

# Eliminar filas con valores ausentes
auto = auto.dropna().copy()

print("\nFilas eliminadas:", filas_antes - len(auto))
print("Filas restantes:", len(auto))
print("Valores ausentes después:", auto.isna().sum().sum())
```

Find the frequency table for the `cylinders` column using the `value_counts()` function. How many possible values of cylinders are there?

```python

# Contar coches por número de cilindros
frecuencia = auto["cylinders"].value_counts().sort_index()

print("Frecuencia de cilindros:")
print(frecuencia)

print("\nCantidad de valores diferentes:",
      auto["cylinders"].nunique())
```

We would like to generate a linear regression model that will predict mpg. To do this, first drop the `car_name` column since it does not contain any quantitative data. Next separate the dataframe to predictor and response variables. Separate those into test and training data with 80% of the data in the training set and the remainder in the test set.

Assign the predictor and response training data to `X_train` and `y_train` respectively. Similarly, assign the predictor and response test data to `X_test` and `y_test`.

```python

# Separar las variables predictoras
X = auto.drop(columns=["car_name", "mpg"])

# Variable que queremos predecir
y = auto["mpg"]

# Dividir los datos: 80% entrenamiento, 20% prueba
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    train_size=0.80,
    random_state=42
)

print("Entrenamiento:", X_train.shape, y_train.shape)
print("Prueba:", X_test.shape, y_test.shape)

# Verificar que no hay datos ausentes
print("NaN en entrenamiento:", X_train.isna().sum().sum())
print("NaN en prueba:", X_test.isna().sum().sum())
```

Now we will the dataset that we processed and peform linear regression on this data to predict the mpg for each vehicle. Initialize the model in the cell below.

```python

# Crear el modelo de regresión lineal
auto_model = LinearRegression()

print("Modelo de regresión lineal creado correctamente")

```

Next, fit the model in the cell below.

```python

# Entrenar el modelo
auto_model.fit(X_train, y_train)

print("Modelo entrenado correctamente")

# Mostrar los resultados del modelo
print("Intercepto:", auto_model.intercept_)
print("Coeficientes:", auto_model.coef_)
```

# Challenge 4 - Evaluate the Model

the r squared score of a model tells us how much variation is explained by the model. In a typical dataset, most observations differ from the mean. When we create a model, we are trying to generate an equation that will tell us by how much each observation will differ from the mean. Obviously, the vast majority of models are not perfect. They can only predict some of the variation from the mean but not all of it. We attribute the rest of the difference between the actual value and the mean to random error. We would like random error to explain the as little as possible of the variation. This is why the r squared score is an important metric.

In the next cell, compute the r squared score of the model. Do this by first computing the predicted values and assign them to `y_pred`.

```python

from sklearn.metrics import r2_score

# Realizar predicciones con los datos de entrenamiento
y_pred = auto_model.predict(X_train)

# Calcular R²
r2_train = r2_score(y_train, y_pred)

print("R² del modelo (entrenamiento):", r2_train)
```

#### Our next step is to evaluate the model using the test data. We would like to ensure that our model is not overfitting the data. This means that our model will not be able to generalize well outside of the training data.

In the cell below, use the model to generate the predicted values for the training data and assign them to `y_test_pred`. Compute the r squared score for the test data by comparing the oberserved `y_train` data and the predicted `y_test_pred`.

```python

# Predecir utilizando los datos de prueba
y_test_pred = auto_model.predict(X_test)

# Calcular R² con los datos de prueba
r2_test = r2_score(y_test, y_test_pred)

print("R² entrenamiento:", r2_train)
print("R² prueba:", r2_test)

# Comprobar la diferencia
print("Diferencia:", r2_train - r2_test)
```

# Challenge 5 - Improve the Model Fit

While the most common way to improve the fit of a model is by using regularization, there are other simpler ways to improve model fit. The first is to create a simpler model. The second is to increase the train sample size.

Let us start with the easier option and increase our train sample size to 90% of the data. Create a new test train split and name the new predictors and response variables `X_train09`, `X_test09`, `y_train09`, `y_test09`.

```python

from sklearn.model_selection import train_test_split

# Crear una nueva división: 90% entrenamiento
X_train09, X_test09, y_train09, y_test09 = train_test_split(
    X,
    y,
    train_size=0.90,
    random_state=42
)

print("Entrenamiento:", X_train09.shape, y_train09.shape)
print("Prueba:", X_test09.shape, y_test09.shape)
```

Initialize a new model. Name this model `auto_model09`. Fit the model to the new sample data.

```python

from sklearn.linear_model import LinearRegression

# Crear un nuevo modelo
auto_model09 = LinearRegression()

# Entrenar el modelo con el 90% de los datos
auto_model09.fit(X_train09, y_train09)

print("Nuevo modelo entrenado correctamente")
```

Compute the predicted values and r squared score for our new model and new sample data.

```python

# Realizar predicciones en el entrenamiento
y_pred09 = auto_model09.predict(X_train09)

# Calcular R²
r2_train09 = r2_score(y_train09, y_pred09)

print("R² del nuevo modelo (90% entrenamiento):", r2_train09)

# Comparar con el modelo anterior
print("R² modelo anterior:", r2_train)
print("R² modelo nuevo:", r2_train09)
```

Compute the r squared score for the smaller test set. Is there an improvement in the test r squared?

```python

# Predecir con los datos de prueba del nuevo modelo
y_test_pred09 = auto_model09.predict(X_test09)

# Calcular R² de prueba
r2_test09 = r2_score(y_test09, y_test_pred09)

print("R² prueba modelo 80%:", r2_test)
print("R² prueba modelo 90%:", r2_test09)

# Comparar resultados
if r2_test09 > r2_test:
    print("El nuevo modelo tiene un R² de prueba mayor.")
elif r2_test09 < r2_test:
    print("El nuevo modelo tiene un R² de prueba menor.")
else:
    print("Los dos modelos tienen el mismo R² de prueba.")
```

# Bonus Challenge 2 - Backward Elimination

The main way to produce a simpler linear regression model is to reduce the number of variables used in the model. In scikit-learn, we can do this by using recursive feature elimination. You can read more about RFE [here](https://scikit-learn.org/stable/modules/generated/sklearn.feature_selection.RFE.html).

In the next cell, we will import RFE

```python
from sklearn.feature_selection import RFE
```

Follow the documentation and initialize an RFE model using the `auto_model` linear regression model. Set `n_features_to_select=3`

```python

# Crear el selector de variables
rfe = RFE(
    estimator=LinearRegression(),
    n_features_to_select=3
)

print("Modelo RFE inicializado correctamente")
```

Fit the model and print the ranking

```python

# Entrenar el selector RFE
rfe.fit(X_train, y_train)

# Mostrar los resultados
ranking = pd.DataFrame({
    "Variable": X_train.columns,
    "Ranking": rfe.ranking_,
    "Seleccionada": rfe.support_
})

display(ranking.sort_values("Ranking"))
```

Feature importance is ranked from most important (1) to least important (4). Generate a model with the three most important features. The features correspond to variable names. For example, feature 1 is `cylinders` and feature 2 is `displacement`.

Perform a test-train split on this reduced column data and call the split data `X_train_reduced`, `X_test_reduced`, `y_test_reduced`, `y_train_reduced`. Use an 80% split.

```python

# Obtener las tres variables seleccionadas
selected_features = X_train.columns[rfe.support_].tolist()

print("Variables seleccionadas:", selected_features)

# Crear un DataFrame reducido
X_reduced = X[selected_features]

# Dividir el conjunto reducido: 80% y 20%
X_train_reduced, X_test_reduced, y_train_reduced, y_test_reduced = train_test_split(
    X_reduced,
    y,
    train_size=0.80,
    random_state=42
)

print("Entrenamiento:", X_train_reduced.shape)
print("Prueba:", X_test_reduced.shape)
```

Generate a new model called `auto_model_reduced` and fit this model. Then proceed to compute the r squared score for the model. Did this cause an improvement in the r squared score?

```python

# Crear el modelo reducido
auto_model_reduced = LinearRegression()

# Entrenar con las tres variables seleccionadas
auto_model_reduced.fit(X_train_reduced, y_train_reduced)

# Predicciones del entrenamiento
y_pred_reduced = auto_model_reduced.predict(X_train_reduced)

# Predicciones de prueba
y_test_pred_reduced = auto_model_reduced.predict(X_test_reduced)

# Calcular R²
r2_train_reduced = r2_score(y_train_reduced, y_pred_reduced)
r2_test_reduced = r2_score(y_test_reduced, y_test_pred_reduced)

print("R² entrenamiento modelo original:", r2_train)
print("R² entrenamiento modelo reducido:", r2_train_reduced)

print("R² prueba modelo original:", r2_test)
print("R² prueba modelo reducido:", r2_test_reduced)

# Comparar el rendimiento de prueba
if r2_test_reduced > r2_test:
    print("El modelo reducido tiene un R² de prueba mayor.")
elif r2_test_reduced < r2_test:
    print("El modelo reducido tiene un R² de prueba menor.")
else:
    print("Ambos modelos tienen el mismo R² de prueba.")
```
